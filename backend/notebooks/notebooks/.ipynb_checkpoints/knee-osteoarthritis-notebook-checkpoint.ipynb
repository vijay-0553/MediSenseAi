# MediSense AI — Knee OA EfficientNet-B5 + CleanLab

**Purpose:** Train a 5-class Kellgren–Lawrence knee osteoarthritis severity model in Kaggle using GPU, following the methodology of:

**Momenpour & Abu Mallouh (Diagnostics, 2025): "Optimizing CNN-Based Diagnosis of Knee Osteoarthritis: Enhancing Model Accuracy with CleanLab Relabeling."**

Target reported result: **82.07% accuracy** and **80.19% macro F1** on the paper's CleanLab-remediated test labels.

### Important dataset interpretation

Your current Kaggle dataset contains:

- `train/` = **5,778** images
- `auto_test/` = **1,526** images
- `val/` = **826** images
- `test/` = **1,656** images

This is important: **do NOT merge `auto_test` back into `train`.**

The paper starts with 7,304 training images and performs outlier removal. The paper then reports CleanLab relabeling on **5,778** remaining training examples. Your current `train + auto_test = 7,304`, and the exact 1,526-image `auto_test` partition therefore corresponds to the outlier-removal stage already present in your dataset.

This notebook therefore treats:

`train/` → outlier-removed training set used by CleanLab

`auto_test/` → retained only as an audit of the previously removed outliers

`val/` → validation set

`test/` → test set

The notebook does **not** copy or move any source images.

````python
# ============================================================
# 2. CONFIGURATION
# ============================================================

SEED = 42

NUM_CLASSES = 5
CLASS_NAMES = ["KL-0", "KL-1", "KL-2", "KL-3", "KL-4"]

IMAGE_SIZE = 224

# Kaggle T4/P100-safe default.
# The paper allowed a dynamically selected batch size up to 80.
BATCH_SIZE = 32

NUM_WORKERS = 2

# Paper: maximum 50 epochs.
MAX_EPOCHS_PRELIM = 50
MAX_EPOCHS_FINAL = 50

# Paper: Adam, base LR = 1e-4.
LEARNING_RATE = 1e-4

# Paper: LR reduced by factor 0.2; minimum LR 1e-6.
LR_FACTOR = 0.2
MIN_LR = 1e-6

# Paper: stop after 3 consecutive epochs without validation-loss improvement.
PATIENCE = 3

# Paper custom head:
# GAP -> Dropout(0.5) -> Dense(256) -> BN -> Dropout(0.4) -> 5-class Softmax
HEAD_DROPOUT_1 = 0.50
HEAD_DROPOUT_2 = 0.40
HEAD_HIDDEN = 256
L2_WEIGHT_DECAY = 1e-4

# Outlier stage:
# The current dataset already contains the paper-like 5,778/1,526 split.
USE_EXISTING_OUTLIER_SPLIT = True

# If you later have the untouched 7,304-image training pool and want
# to recompute outliers, set this False and use the optional AE section.
RECOMPUTE_OUTLIERS = False

# The paper uses CleanLab on train, validation and test.
RUN_CLEANLAB_ON_VAL = True
RUN_CLEANLAB_ON_TEST = True

# Final model and artifacts
OUTPUT_DIR = "/kaggle/working/mediSense_knee_outputs"

# Optional extras
SAVE_GRADCAM = True

import os, random, json, time, math, copy
os.makedirs(OUTPUT_DIR, exist_ok=True)

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

print("Configuration loaded.")
````

````python
# ============================================================
# 3. DATASET DISCOVERY
# ============================================================

from pathlib import Path

INPUT_ROOT = Path("/kaggle/input")

def find_dataset_root(root):
    candidates = []

    for p in [root] + [x for x in root.rglob("*") if x.is_dir()]:
        try:
            child_dirs = {x.name.lower() for x in p.iterdir() if x.is_dir()}
        except Exception:
            continue

        if {"train", "val", "test"}.issubset(child_dirs):
            candidates.append(p)

    if not candidates:
        raise FileNotFoundError(
            "Could not find a dataset containing train/, val/, and test/ under /kaggle/input."
        )

    # Prefer the dataset containing auto_test because this is your current layout.
    for p in candidates:
        if (p / "auto_test").exists():
            return p

    return candidates[0]

DATASET_DIR = find_dataset_root(INPUT_ROOT)

TRAIN_DIR = DATASET_DIR / "train"
AUTO_TEST_DIR = DATASET_DIR / "auto_test"
VAL_DIR = DATASET_DIR / "val"
TEST_DIR = DATASET_DIR / "test"

EXTS = {".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff"}

def image_files(directory):
    directory = Path(directory)
    if not directory.exists():
        return []
    return sorted([
        p for p in directory.rglob("*")
        if p.is_file() and p.suffix.lower() in EXTS
    ])

def split_counts(directory):
    return {
        c: len(image_files(Path(directory) / str(c)))
        for c in range(NUM_CLASSES)
    }

train_counts = split_counts(TRAIN_DIR)
auto_counts = split_counts(AUTO_TEST_DIR)
val_counts = split_counts(VAL_DIR)
test_counts = split_counts(TEST_DIR)

print("=" * 70)
print("DATASET FOUND")
print("=" * 70)
print("Dataset root:", DATASET_DIR)
print()
print("Train:", train_counts, "TOTAL:", sum(train_counts.values()))
print("Auto-test:", auto_counts, "TOTAL:", sum(auto_counts.values()))
print("Validation:", val_counts, "TOTAL:", sum(val_counts.values()))
print("Test:", test_counts, "TOTAL:", sum(test_counts.values()))

assert sum(train_counts.values()) == 5778
assert sum(auto_counts.values()) == 1526
assert sum(val_counts.values()) == 826
assert sum(test_counts.values()) == 1656

assert sum(train_counts.values()) + sum(auto_counts.values()) == 7304
assert sum(train_counts.values()) + sum(auto_counts.values()) + sum(val_counts.values()) + sum(test_counts.values()) == 9786

print()
print("PASS")
print("7,304 original training pool = 5,778 retained train + 1,526 auto-test/outliers")
print("Total dataset = 9,786 images")
````

## 4. Why `auto_test` is NOT merged

This is the correction to the previous notebook.

The research paper says the outlier stage leaves **5,778 training images** for the CleanLab-input model. It later reports **666 relabeled images out of 5,778**.

Your dataset has exactly:

- 5,778 in `train`
- 1,526 in `auto_test`

Therefore the 1,526 images are the already-removed outlier pool. Merging them back would invalidate the intended preprocessing sequence.

````python
# ============================================================
# 5. BUILD DATAFRAME
# ============================================================

rows = []

def add_split(directory, split_name):
    for label in range(NUM_CLASSES):
        for p in image_files(Path(directory) / str(label)):
            rows.append({
                "path": str(p),
                "label": int(label),
                "split": split_name
            })

add_split(TRAIN_DIR, "train")
add_split(VAL_DIR, "val")
add_split(TEST_DIR, "test")

df = pd.DataFrame(rows)

print("Images used by CleanLab/model pipeline:", len(df))
print()
print(pd.crosstab(df["split"], df["label"]))
print()

assert len(df) == 8260
assert (df["split"] == "train").sum() == 5778
assert (df["split"] == "val").sum() == 826
assert (df["split"] == "test").sum() == 1656

print("PASS: model dataframe contains 5,778 + 826 + 1,656 = 8,260 images.")
````

````python
# ============================================================
# 6. IMAGE INTEGRITY CHECK
# ============================================================

from PIL import Image

bad_images = []
blank_images = []
dimensions = {}

for i, p in enumerate(df["path"]):
    try:
        with Image.open(p) as img:
            img.verify()

        with Image.open(p) as img:
            arr = np.asarray(img.convert("L"))
            dimensions[arr.shape] = dimensions.get(arr.shape, 0) + 1

            if arr.std() < 1e-6:
                blank_images.append(p)

    except Exception as e:
        bad_images.append((p, str(e)))

print("Total:", len(df))
print("Corrupt:", len(bad_images))
print("Blank:", len(blank_images))
print("Dimensions:", dimensions)

assert len(bad_images) == 0
assert len(blank_images) == 0

print("PASS: all model images are readable and non-blank.")
````

````python
# ============================================================
# 8. PYTORCH DATASETS + TRANSFORMS
# ============================================================

from torch.utils.data import Dataset, DataLoader
from torchvision import transforms

class KneeDataset(Dataset):
    def __init__(self, frame, transform=None):
        self.frame = frame.reset_index(drop=True).copy()
        self.transform = transform

    def __len__(self):
        return len(self.frame)

    def __getitem__(self, idx):
        row = self.frame.iloc[idx]

        img = Image.open(row["path"]).convert("RGB")
        label = int(row["label"])

        if self.transform:
            img = self.transform(img)

        return img, label, row["path"]


# Paper: horizontal flip + rotation ±10 degrees.
train_transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomRotation(degrees=10),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

eval_transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

train_df = df[df["split"] == "train"].copy()
val_df = df[df["split"] == "val"].copy()
test_df = df[df["split"] == "test"].copy()

train_ds = KneeDataset(train_df, train_transform)
val_ds = KneeDataset(val_df, eval_transform)
test_ds = KneeDataset(test_df, eval_transform)

train_loader = DataLoader(
    train_ds, batch_size=BATCH_SIZE, shuffle=True,
    num_workers=NUM_WORKERS, pin_memory=True
)

val_loader = DataLoader(
    val_ds, batch_size=BATCH_SIZE, shuffle=False,
    num_workers=NUM_WORKERS, pin_memory=True
)

test_loader = DataLoader(
    test_ds, batch_size=BATCH_SIZE, shuffle=False,
    num_workers=NUM_WORKERS, pin_memory=True
)

print("Train batches:", len(train_loader))
print("Val batches:", len(val_loader))
print("Test batches:", len(test_loader))
````

````python
# ============================================================
# 9. EFFICIENTNET-B5 MODEL
# ============================================================

import torch.nn as nn
from torchvision.models import efficientnet_b5, EfficientNet_B5_Weights

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

class EfficientNetB5_Knee(nn.Module):
    def __init__(self, num_classes=5, pretrained=True):
        super().__init__()

        weights = EfficientNet_B5_Weights.IMAGENET1K_V1 if pretrained else None
        self.backbone = efficientnet_b5(weights=weights)

        # Remove original ImageNet classifier.
        in_features = self.backbone.classifier[1].in_features
        self.backbone.classifier = nn.Identity()

        # Paper-inspired custom head:
        # GAP is already performed by EfficientNet's forward path.
        self.dropout1 = nn.Dropout(HEAD_DROPOUT_1)
        self.fc = nn.Linear(in_features, HEAD_HIDDEN)
        self.bn = nn.BatchNorm1d(HEAD_HIDDEN)
        self.relu = nn.ReLU(inplace=True)
        self.dropout2 = nn.Dropout(HEAD_DROPOUT_2)
        self.out = nn.Linear(HEAD_HIDDEN, num_classes)

    def forward_features(self, x):
        x = self.backbone.features(x)
        x = self.backbone.avgpool(x)
        x = torch.flatten(x, 1)
        return x

    def forward(self, x, return_embedding=False):
        x = self.forward_features(x)

        embedding = self.dropout1(x)

        x = self.fc(embedding)
        x = self.bn(x)
        x = self.relu(x)
        x = self.dropout2(x)

        logits = self.out(x)

        if return_embedding:
            return logits, x

        return logits


model_check = EfficientNetB5_Knee(pretrained=True).to(DEVICE)

print("Device:", DEVICE)
print("Parameters:", sum(p.numel() for p in model_check.parameters()))
print(model_check)
````

## 10. Preliminary EfficientNet-B5 — CleanLab input model

The paper trains an initial EfficientNet-B5 on the **outlier-removed training set** specifically to generate:

1. predicted class probabilities
2. penultimate-layer feature embeddings

for CleanLab.

The paper states that this preliminary model uses standard categorical cross-entropy **without class weights**.

This notebook follows that rule.

````python
# ============================================================
# 10. TRAINING UTILITIES
# ============================================================

from sklearn.metrics import accuracy_score, precision_recall_fscore_support, classification_report, confusion_matrix

def evaluate_model(model, loader, device=DEVICE):
    model.eval()

    all_labels = []
    all_preds = []
    all_probs = []
    all_embeddings = []

    with torch.no_grad():
        for images, labels, paths in loader:
            images = images.to(device, non_blocking=True)

            logits, embeddings = model(images, return_embedding=True)
            probs = torch.softmax(logits, dim=1)

            preds = probs.argmax(dim=1)

            all_labels.extend(labels.numpy().tolist())
            all_preds.extend(preds.cpu().numpy().tolist())
            all_probs.append(probs.cpu().numpy())
            all_embeddings.append(embeddings.cpu().numpy())

    all_probs = np.concatenate(all_probs)
    all_embeddings = np.concatenate(all_embeddings)

    acc = accuracy_score(all_labels, all_preds)
    precision, recall, f1, _ = precision_recall_fscore_support(
        all_labels, all_preds,
        labels=list(range(NUM_CLASSES)),
        average="macro",
        zero_division=0
    )

    return {
        "labels": np.array(all_labels),
        "preds": np.array(all_preds),
        "probs": all_probs,
        "embeddings": all_embeddings,
        "accuracy": acc,
        "macro_precision": precision,
        "macro_recall": recall,
        "macro_f1": f1
    }


def train_model(
    model,
    train_loader,
    val_loader,
    max_epochs,
    class_weights=None,
    name="model"
):
    model = model.to(DEVICE)

    criterion = nn.CrossEntropyLoss(
        weight=class_weights.to(DEVICE) if class_weights is not None else None
    )

    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=LEARNING_RATE,
        weight_decay=L2_WEIGHT_DECAY
    )

    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode="min",
        factor=LR_FACTOR,
        patience=PATIENCE,
        min_lr=MIN_LR
    )

    scaler = torch.amp.GradScaler("cuda", enabled=torch.cuda.is_available())

    history = []
    best_state = None
    best_val_loss = float("inf")
    epochs_without_improvement = 0

    for epoch in range(1, max_epochs + 1):

        model.train()

        running_loss = 0.0
        correct = 0
        total = 0

        start = time.time()

        for images, labels, paths in train_loader:

            images = images.to(DEVICE, non_blocking=True)
            labels = labels.to(DEVICE, non_blocking=True)

            optimizer.zero_grad(set_to_none=True)

            with torch.amp.autocast(
                device_type="cuda",
                enabled=torch.cuda.is_available()
            ):
                logits = model(images)
                loss = criterion(logits, labels)

            scaler.scale(loss).backward()
            scaler.step(optimizer)
            scaler.update()

            running_loss += loss.item() * labels.size(0)
            correct += (logits.argmax(1) == labels).sum().item()
            total += labels.size(0)

        train_loss = running_loss / total
        train_acc = correct / total

        # Validation
        model.eval()

        val_loss_total = 0.0
        val_total = 0
        val_correct = 0

        with torch.no_grad():
            for images, labels, paths in val_loader:

                images = images.to(DEVICE, non_blocking=True)
                labels = labels.to(DEVICE, non_blocking=True)

                logits = model(images)
                loss = criterion(logits, labels)

                val_loss_total += loss.item() * labels.size(0)
                val_correct += (logits.argmax(1) == labels).sum().item()
                val_total += labels.size(0)

        val_loss = val_loss_total / val_total
        val_acc = val_correct / val_total

        scheduler.step(val_loss)

        current_lr = optimizer.param_groups[0]["lr"]

        elapsed = time.time() - start

        history.append({
            "epoch": epoch,
            "train_loss": train_loss,
            "train_acc": train_acc,
            "val_loss": val_loss,
            "val_acc": val_acc,
            "lr": current_lr
        })

        print(
            f"[{name}] Epoch {epoch:02d}/{max_epochs} | "
            f"train_loss={train_loss:.4f} | "
            f"train_acc={train_acc:.4f} | "
            f"val_loss={val_loss:.4f} | "
            f"val_acc={val_acc:.4f} | "
            f"lr={current_lr:.2e} | "
            f"{elapsed/60:.1f} min"
        )

        if val_loss < best_val_loss - 1e-5:

            best_val_loss = val_loss
            best_state = copy.deepcopy(model.state_dict())
            epochs_without_improvement = 0

            torch.save(
                best_state,
                os.path.join(OUTPUT_DIR, f"{name}_best_state.pth")
            )

        else:
            epochs_without_improvement += 1

        # Stop once LR reaches minimum and validation loss has stopped improving.
        if current_lr <= MIN_LR + 1e-12 and epochs_without_improvement >= PATIENCE:
            print("Early stopping: minimum learning rate reached.")
            break

    if best_state is not None:
        model.load_state_dict(best_state)

    return model, pd.DataFrame(history)
````

````python
# ============================================================
# 11. PRELIMINARY / CLEANLAB-INPUT MODEL
# ============================================================

prelim_model = EfficientNetB5_Knee(
    num_classes=NUM_CLASSES,
    pretrained=True
).to(DEVICE)

# IMPORTANT:
# No class weights for the CleanLab-input model, matching the paper.
prelim_model, prelim_history = train_model(
    prelim_model,
    train_loader,
    val_loader,
    max_epochs=MAX_EPOCHS_PRELIM,
    class_weights=None,
    name="cleanlab_input_efficientnet_b5"
)

prelim_history.to_csv(
    os.path.join(OUTPUT_DIR, "preliminary_training_history.csv"),
    index=False
)

print("Preliminary model training complete.")
````

````python
# ============================================================
# 12. GENERATE CLEANLAB INPUTS
# ============================================================

# The paper uses the trained CleanLab-input EfficientNet-B5
# to generate predictions and penultimate embeddings for
# train, validation and test.

train_eval_loader = DataLoader(
    KneeDataset(train_df, eval_transform),
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=True
)

val_eval_loader = DataLoader(
    KneeDataset(val_df, eval_transform),
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=True
)

test_eval_loader = DataLoader(
    KneeDataset(test_df, eval_transform),
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=True
)

cleanlab_train = evaluate_model(prelim_model, train_eval_loader)
cleanlab_val = evaluate_model(prelim_model, val_eval_loader)
cleanlab_test = evaluate_model(prelim_model, test_eval_loader)

print("CleanLab input shapes:")
print("Train probabilities:", cleanlab_train["probs"].shape)
print("Train embeddings:", cleanlab_train["embeddings"].shape)
print("Val probabilities:", cleanlab_val["probs"].shape)
print("Test probabilities:", cleanlab_test["probs"].shape)

print()
print("Preliminary train accuracy:", cleanlab_train["accuracy"])
print("Preliminary val accuracy:", cleanlab_val["accuracy"])
print("Preliminary test accuracy:", cleanlab_test["accuracy"])
````

````python
# ============================================================
# 13. CLEANLAB LABEL AUDIT
# ============================================================

from cleanlab import Datalab

def run_cleanlab(labels, pred_probs, features, split_name):
    print("=" * 70)
    print("CLEANLAB:", split_name)
    print("=" * 70)

    data = {
        "X": np.arange(len(labels)),
        "label": labels.astype(int)
    }

    lab = Datalab(
        data=data,
        label_name="label"
    )

    # We explicitly request label issues.
    # This keeps the audit focused on the paper's relabeling objective.
    lab.find_issues(
        pred_probs=pred_probs,
        features=features,
        issue_types={"label": {}}
    )

    issues = lab.get_issues("label").copy()

    # CleanLab provides the original label and suggested predicted_label.
    suggested = issues["predicted_label"].astype(int).to_numpy()

    changed = labels.astype(int) != suggested

    issues["original_label"] = labels.astype(int)
    issues["changed"] = changed

    print("Total samples:", len(labels))
    print("Potential label issues:", int(changed.sum()))
    print()

    print("Relabel transition matrix:")
    print(
        pd.crosstab(
            labels.astype(int),
            suggested,
            rownames=["Original"],
            colnames=["CleanLab suggested"]
        )
    )

    return lab, issues, suggested


train_lab, train_issues, train_clean_labels = run_cleanlab(
    cleanlab_train["labels"],
    cleanlab_train["probs"],
    cleanlab_train["embeddings"],
    "TRAIN"
)

if RUN_CLEANLAB_ON_VAL:
    val_lab, val_issues, val_clean_labels = run_cleanlab(
        cleanlab_val["labels"],
        cleanlab_val["probs"],
        cleanlab_val["embeddings"],
        "VALIDATION"
    )
else:
    val_clean_labels = val_df["label"].to_numpy()

if RUN_CLEANLAB_ON_TEST:
    test_lab, test_issues, test_clean_labels = run_cleanlab(
        cleanlab_test["labels"],
        cleanlab_test["probs"],
        cleanlab_test["embeddings"],
        "TEST"
    )
else:
    test_clean_labels = test_df["label"].to_numpy()

print()
print("Paper-reported reference counts:")
print("Train relabeled: 666 / 5778")
print("Val relabeled:   215 / 826")
print("Test relabeled:  366 / 1656")
print()
print("Our CleanLab counts:")
print("Train:", int((train_clean_labels != train_df["label"].to_numpy()).sum()))
print("Val:  ", int((val_clean_labels != val_df["label"].to_numpy()).sum()))
print("Test: ", int((test_clean_labels != test_df["label"].to_numpy()).sum()))
````

### CleanLab interpretation

The paper reports:

- **666** train labels changed out of 5,778
- **215** validation labels changed out of 826
- **366** test labels changed out of 1,656

Your run may not produce exactly these numbers because model initialization, framework implementation, preprocessing, package versions, and training convergence can differ. The counts are a useful **methodology checkpoint**, not a requirement that must be forced.

````python
# ============================================================
# 14. SAVE CLEANLAB LABEL TABLES
# ============================================================

train_clean_df = train_df.copy()
train_clean_df["clean_label"] = train_clean_labels
train_clean_df["label_changed"] = (
    train_clean_df["label"] != train_clean_df["clean_label"]
)

val_clean_df = val_df.copy()
val_clean_df["clean_label"] = val_clean_labels
val_clean_df["label_changed"] = (
    val_clean_df["label"] != val_clean_df["clean_label"]
)

test_clean_df = test_df.copy()
test_clean_df["clean_label"] = test_clean_labels
test_clean_df["label_changed"] = (
    test_clean_df["label"] != test_clean_df["clean_label"]
)

train_clean_df.to_csv(
    os.path.join(OUTPUT_DIR, "cleanlab_train_labels.csv"),
    index=False
)

val_clean_df.to_csv(
    os.path.join(OUTPUT_DIR, "cleanlab_val_labels.csv"),
    index=False
)

test_clean_df.to_csv(
    os.path.join(OUTPUT_DIR, "cleanlab_test_labels.csv"),
    index=False
)

train_issues.to_csv(
    os.path.join(OUTPUT_DIR, "cleanlab_train_issues.csv"),
    index=False
)

if RUN_CLEANLAB_ON_VAL:
    val_issues.to_csv(
        os.path.join(OUTPUT_DIR, "cleanlab_val_issues.csv"),
        index=False
    )

if RUN_CLEANLAB_ON_TEST:
    test_issues.to_csv(
        os.path.join(OUTPUT_DIR, "cleanlab_test_issues.csv"),
        index=False
    )

print("CleanLab artifacts saved.")
````

````python
# ============================================================
# 15. CLEANLAB LABEL DISTRIBUTIONS
# ============================================================

def distribution(frame, column):
    return frame[column].value_counts().sort_index()

print("Original training distribution:")
print(distribution(train_clean_df, "label"))

print("CleanLab training distribution:")
print(distribution(train_clean_df, "clean_label"))

print("Original validation distribution:")
print(distribution(val_clean_df, "label"))

print("CleanLab validation distribution:")
print(distribution(val_clean_df, "clean_label"))

print("Original test distribution:")
print(distribution(test_clean_df, "label"))

print("CleanLab test distribution:")
print(distribution(test_clean_df, "clean_label"))
````

````python
# ============================================================
# 16. FINAL DATASETS USING CLEANLAB LABELS
# ============================================================

class CleanLabelDataset(Dataset):
    def __init__(self, frame, label_column="clean_label", transform=None):
        self.frame = frame.reset_index(drop=True).copy()
        self.label_column = label_column
        self.transform = transform

    def __len__(self):
        return len(self.frame)

    def __getitem__(self, idx):
        row = self.frame.iloc[idx]

        img = Image.open(row["path"]).convert("RGB")
        label = int(row[self.label_column])

        if self.transform:
            img = self.transform(img)

        return img, label, row["path"]


final_train_ds = CleanLabelDataset(
    train_clean_df,
    label_column="clean_label",
    transform=train_transform
)

final_val_ds = CleanLabelDataset(
    val_clean_df,
    label_column="clean_label",
    transform=eval_transform
)

final_test_ds = CleanLabelDataset(
    test_clean_df,
    label_column="clean_label",
    transform=eval_transform
)

final_train_loader = DataLoader(
    final_train_ds,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=NUM_WORKERS,
    pin_memory=True
)

final_val_loader = DataLoader(
    final_val_ds,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=True
)

final_test_loader = DataLoader(
    final_test_ds,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=True
)

print("Final CleanLab training samples:", len(final_train_ds))
print("Final CleanLab validation samples:", len(final_val_ds))
print("Final CleanLab test samples:", len(final_test_ds))
````

````python
# ============================================================
# 17. CLASS WEIGHTS — FINAL MODEL
# ============================================================

# Paper: inverse-frequency class weights computed from the
# CleanLab-remediated training set.

clean_counts = (
    train_clean_df["clean_label"]
    .value_counts()
    .sort_index()
    .reindex(range(NUM_CLASSES), fill_value=0)
    .to_numpy()
)

inverse_weights = 1.0 / np.maximum(clean_counts, 1)

# Normalize so average weight is approximately 1.
inverse_weights = inverse_weights / inverse_weights.mean()

class_weights = torch.tensor(
    inverse_weights,
    dtype=torch.float32
)

print("CleanLab training counts:")
for c, n in enumerate(clean_counts):
    print(f"KL-{c}: {n}")

print()
print("Class weights:")
for c, w in enumerate(class_weights.numpy()):
    print(f"KL-{c}: {w:.4f}")
````

## 18. Final EfficientNet-B5

The final model is trained **from scratch in terms of model initialization**, using ImageNet-pretrained EfficientNet-B5 weights as the starting backbone, but it is **not initialized from the CleanLab-input model**.

The paper's final training stage uses:

- CleanLab-remediated training labels
- CleanLab-remediated validation labels
- inverse-frequency class weights
- Adam
- learning rate `1e-4`
- horizontal flip
- rotation ±10°
- maximum 50 epochs
- validation-loss based LR reduction
- factor `0.2`
- minimum LR `1e-6`
- early stopping
- EfficientNet-B5 + custom classifier head

````python
# ============================================================
# 18. TRAIN FINAL MODEL
# ============================================================

final_model = EfficientNetB5_Knee(
    num_classes=NUM_CLASSES,
    pretrained=True
).to(DEVICE)

final_model, final_history = train_model(
    final_model,
    final_train_loader,
    final_val_loader,
    max_epochs=MAX_EPOCHS_FINAL,
    class_weights=class_weights,
    name="final_efficientnet_b5_cleanlab"
)

final_history.to_csv(
    os.path.join(OUTPUT_DIR, "final_training_history.csv"),
    index=False
)

print("FINAL MODEL TRAINING COMPLETE.")
````

````python
# ============================================================
# 19. FINAL EVALUATION
# ============================================================

final_results = evaluate_model(
    final_model,
    final_test_loader
)

original_test_labels = test_df["label"].to_numpy()
clean_test_labels = test_clean_df["clean_label"].to_numpy()

preds = final_results["preds"]

original_acc = accuracy_score(original_test_labels, preds)
clean_acc = accuracy_score(clean_test_labels, preds)

orig_p, orig_r, orig_f1, _ = precision_recall_fscore_support(
    original_test_labels,
    preds,
    labels=list(range(NUM_CLASSES)),
    average="macro",
    zero_division=0
)

clean_p, clean_r, clean_f1, _ = precision_recall_fscore_support(
    clean_test_labels,
    preds,
    labels=list(range(NUM_CLASSES)),
    average="macro",
    zero_division=0
)

print("=" * 70)
print("FINAL TEST RESULTS")
print("=" * 70)

print("Against ORIGINAL test labels:")
print(f"Accuracy:    {original_acc*100:.4f}%")
print(f"Macro F1:    {orig_f1*100:.4f}%")
print(f"Macro Prec.: {orig_p*100:.4f}%")
print(f"Macro Recall:{orig_r*100:.4f}%")

print("Against CLEANLAB test labels:")
print(f"Accuracy:    {clean_acc*100:.4f}%")
print(f"Macro F1:    {clean_f1*100:.4f}%")
print(f"Macro Prec.: {clean_p*100:.4f}%")
print(f"Macro Recall:{clean_r*100:.4f}%")

print("Paper reference:")
print("Accuracy: 82.07%")
print("Macro F1: 80.19%")
````

````python
# ============================================================
# 22. GRAD-CAM — FINAL MODEL
# ============================================================

# A lightweight Grad-CAM implementation for the last EfficientNet feature block.

class GradCAM:
    def __init__(self, model):
        self.model = model
        self.activations = None
        self.gradients = None

        target_layer = self.model.backbone.features[-1]
        target_layer.register_forward_hook(self.forward_hook)
        target_layer.register_full_backward_hook(self.backward_hook)

    def forward_hook(self, module, inp, output):
        self.activations = output.detach()

    def backward_hook(self, module, grad_input, grad_output):
        self.gradients = grad_output[0].detach()

    def generate(self, image_tensor, target_class=None):

        self.model.zero_grad(set_to_none=True)

        logits = self.model(image_tensor)

        if target_class is None:
            target_class = logits.argmax(dim=1).item()

        score = logits[:, target_class].sum()
        score.backward()

        gradients = self.gradients
        activations = self.activations

        weights = gradients.mean(dim=(2, 3), keepdim=True)

        cam = (weights * activations).sum(dim=1)
        cam = torch.relu(cam)

        cam = torch.nn.functional.interpolate(
            cam.unsqueeze(1),
            size=(IMAGE_SIZE, IMAGE_SIZE),
            mode="bilinear",
            align_corners=False
        ).squeeze(1)

        cam = cam[0].cpu().numpy()

        cam -= cam.min()
        cam /= (cam.max() + 1e-8)

        return cam, target_class


if SAVE_GRADCAM:

    cam_generator = GradCAM(final_model)

    sample_row = test_clean_df.sample(
        1,
        random_state=SEED
    ).iloc[0]

    raw = Image.open(sample_row["path"]).convert("RGB")
    input_tensor = eval_transform(raw).unsqueeze(0).to(DEVICE)

    cam, predicted_class = cam_generator.generate(input_tensor)

    original = np.asarray(
        raw.resize((IMAGE_SIZE, IMAGE_SIZE))
    )

    plt.figure(figsize=(6, 6))
    plt.imshow(original)
    plt.imshow(cam, cmap="jet", alpha=0.40)
    plt.title(
        f"Grad-CAM | True CleanLab KL-{sample_row['clean_label']} | "
        f"Predicted KL-{predicted_class}"
    )
    plt.axis("off")
    plt.tight_layout()

    plt.savefig(
        os.path.join(OUTPUT_DIR, "gradcam_sample.png"),
        dpi=200,
        bbox_inches="tight"
    )

    plt.show()
````
